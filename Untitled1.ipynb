```python
import time
import sys

# ==========================================
# 1. CURRICULUM KNOWLEDGE GRAPH & DATA BANKS
# ==========================================
KNOWLEDGE_GRAPH = {
    "JEE": {
        "Prerequisites": "Mechanics ➡️ Kinematics ➡️ Newton's Laws",
        "Socratic_Topic": "Conservation of Momentum",
        "Mock_Questions": [
            {"q": "A mass m collides elastically with a mass 2m at rest. What fraction of kinetic energy is transferred?", "options": "A) 1/9\nB) 8/9\nC) 4/9\nD) 2/3", "ans": "B"},
            {"q": "If a particle moves in a circle of radius r with constant speed v, its acceleration is:", "options": "A) Zero\nB) v^2/r towards center\nC) v/r\nD) v^2/r away from center", "ans": "B"}
        ]
    },
    "NEET": {
        "Prerequisites": "Cell Structure ➡️ Biomolecules ➡️ Genetics",
        "Socratic_Topic": "DNA Replication Mechanism",
        "Mock_Questions": [
            {"q": "Which enzyme is responsible for unwinding the DNA double helix?", "options": "A) DNA Polymerase\nB) Helicase\nC) Ligase\nD) RNA Primase", "ans": "B"},
            {"q": "The powerhouse of the cell is known as:", "options": "A) Nucleus\nB) Mitochondria\nC) Ribosome\nD) Golgi Apparatus", "ans": "B"}
        ]
    },
    "CAT": {
        "Prerequisites": "Number Systems ➡️ Arithmetic ➡️ Algebra",
        "Socratic_Topic": "Permutations and Combinations",
        "Mock_Questions": [
            {"q": "In how many ways can the letters of the word 'CAT' be arranged?", "options": "A) 3\nB) 6\nC) 9\nD) 12", "ans": "B"},
            {"q": "If log x + log y = log (x+y), then:", "options": "A) x = y\nB) y = x/(x-1)\nC) y = x/(1-x)\nD) xy = 1", "ans": "B"}
        ]
    }
}

class StudentSession:
    def _init_(self):
        self.name = ""
        self.exam = ""
        self.weekly_goal = 15
        self.completed_hours = 0

session = StudentSession()

# ==========================================
# 2. CORE SYSTEM FEATURES
# ==========================================

def run_diagnostic():
    print("\n=============================================")
    print("📋 MODULE 1: DIAGNOSTIC & STUDY PLANNER")
    print("=============================================")
    session.name = input("Enter Student Name: ").strip() or "Scholar"
    print("\nSelect target exam framework:")
    exams = list(KNOWLEDGE_GRAPH.keys())
    for idx, ex in enumerate(exams, 1):
        print(f" [{idx}] {ex}")

    choice = input("Select choice (1-3): ").strip()
    session.exam = exams[int(choice)-1] if choice in ['1', '2', '3'] else "JEE"
    session.weekly_goal = int(input("Enter targeted weekly study hours (e.g., 15): ") or 15)

    print("\n✨ Generating Study Profile Matrix...")
    time.sleep(1)
    print(f"✔️ Target Exam set to: {session.exam}")
    print(f"✔️ Dynamic Knowledge Dependency Map Loaded: {KNOWLEDGE_GRAPH[session.exam]['Prerequisites']}")
    print(f"📅 7-Day Balanced Timeline Generated successfully for {session.weekly_goal} hours/week!")

def run_socratic():
    print("\n=============================================")
    print("🧠 MODULE 2: SOCRATIC TUTOR ROOM")
    print("=============================================")
    topic = KNOWLEDGE_GRAPH[session.exam]["Socratic_Topic"]
    print(f"Current Topic: {topic}")
    print("AI Tutor: I will not give you direct answers. Let's reason through this step-by-step.")
    print(f"AI Question: What happens to the total system energy when two objects collide without friction?")

    ans = input("Your Answer: ")
    print("\n🤖 AI Evaluating Response Path...")
    time.sleep(1)
    print("AI Hint: You're right that energy can't disappear! But remember to distinguish between absolute total energy and kinetic energy. Think about elastic vs. inelastic constraints. What do you think changes?")

def run_doubt_resolution():
    print("\n=============================================")
    print("📸 MODULE 3: DOUBT RESOLUTION SUB-ENGINE")
    print("=============================================")
    print("[Simulating OCR & Vision Pipeline Scan...]")
    dummy_img = input("Enter name of problem file (e.g., question1.jpg) or press Enter to scan dummy text: ")

    print("\n⚙️ OCR Extracted: 'Solve for x if 3x + 7 = 22'")
    print("🤖 Step-by-Step AI Resolution Matrix:")
    print("   👉 Step 1: Isolate the variable term by subtracting 7 from both sides: 3x = 22 - 7")
    print("   👉 Step 2: Simplify the right-hand equation: 3x = 15")
    print("✨ Solution Confirmed!")

def run_mock():
    print("\n=============================================")
    print("⏱️ MODULE 4: DYNAMIC MOCK SIMULATOR")
    print("=============================================")
    questions = KNOWLEDGE_GRAPH[session.exam]["Mock_Questions"]
    score = 0

    for idx, item in enumerate(questions, 1):
        print(f"\nQ{idx}: {item['q']}\n{item['options']}\n")
        user_ans = input(f"Your Answer for Q{idx} (A/B/C/D): ").strip().upper()
        if user_ans == item['ans']:
            score += 1
            print("✅ Evaluation: Correct!")
        else:
            print(f"❌ Evaluation: Incorrect. (Correct Answer: {item['ans']})")

    print("\n--- Diagnostic Analytics Dashboard ---")
    print(f"🎯 Raw Score: {score}/{len(questions)}")
    if score == len(questions):
        print("📊 Error Analysis: 0 Conceptual Errors detected. High application speed.")
    else:
        print("📊 Error Analysis: Careless calculation patterns identified in foundational variables. Recommended focus: Re-verify calculation matrices.")
    print("📈 National Benchmark Percentile Ranking: 88.4th Percentile among peer test-takers.")

def run_motivation():
    print("\n=============================================")
    print("🌟 MODULE 5: ACCOUNTABILITY & MOTIVATION STATUS")
    print("=============================================")
    session.completed_hours += 2
    progress = (session.completed_hours / session.weekly_goal) * 100
    print(f"👤 Student: {session.name}")
    print(f"📊 Weekly Micro-Milestone Tracker: {session.completed_hours}/{session.weekly_goal} hours completed ({progress:.1f}%)")
    print("🔥 Milestone Achievement Unlocked: Daily Streak Starter!")
    print("\n🤖 AI Mentor: Great job staying committed! Consistency breaks the hardest exam patterns. Keep pushing, you are on track!")

# ==========================================
# 3. INTERACTIVE CONSOLE DRIVER
# ==========================================
def main():
    run_diagnostic()
    while True:
        print("\n" + "="*45)
        print("🌟 ENGINE MAIN MENU 🌟")
        print("="*45)
        print("1. Launch Socratic Classroom Session")
        print("2. Run Image/Text Doubt Resolver")
        print("3. Simulate Live Mock Exam & Error Scan")
        print("4. Pull Accountability Metrics & Motivation Dashboard")
        print("5. Reset/Change Student Profile Settings")
        print("6. Shutdown Platform Engine")

        choice = input("\nSelect system module to execute (1-6): ").strip()
        if choice == '1':
            run_socratic()
        elif choice == '2':
            run_doubt_resolution()
        elif choice == '3':
            run_mock()
        elif choice == '4':
            run_motivation()
        elif choice == '5':
            run_diagnostic()
        elif choice == '6':
            print(f"\nClosing platform safely. Good luck with your project presentation tonight, {session.name}!")
            break
        else:
            print("⚠️ System Warning: Invalid option. Select a digit between 1 and 6.")

if __name__ == "__main__":
    main()
```
